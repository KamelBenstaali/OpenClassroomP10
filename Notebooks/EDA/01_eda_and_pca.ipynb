## Popularité des articles

Basée sur le nombres de clicks/Age de l'article en mois

```python
# Calcul du nombre d'utilisateurs uniques par article
popularity_df = clicks_df.groupby('click_article_id')['user_id'].nunique().reset_index()
popularity_df.columns = ['click_article_id', 'unique_users_count']

# Tri optionnel pour l'affichage
popularity_df = popularity_df.sort_values('unique_users_count', ascending=False)

# --- Popularité Avancée (Time Decay) ---
print("Calcul de la popularité avec pondération par l'âge (Time Decay)...")

# 1. On charge les dates de création des articles
articles_meta = pd.read_csv('../../Data/news-portal-user-interactions-by-globocom/articles_metadata.csv')
articles_dates = articles_meta[['article_id', 'created_at_ts']]

# 2. On fusionne la popularité classique avec la date de création
popularity_advanced = popularity_df.merge(
    articles_dates, 
    left_on='click_article_id', 
    right_on='article_id', 
    how='left'
)

# On identifie le nom de ta colonne de comptage (ex: 'click_count' ou 'unique_users_count')
count_col = popularity_df.columns[1] 

# 3. Définir le "Présent" = la date de création de l'article le plus récent du dataset
# (Les timestamps sont en millisecondes)
T_max = popularity_advanced['created_at_ts'].max()

# 4. Calcul de l'âge de chaque article en HEURES 
# (Pour éviter la division par zéro sur l'article tout juste sorti, on ajoute 1 heure)
# 1 heure = 3 600 000 millisecondes
popularity_advanced['age_hours'] = ((T_max - popularity_advanced['created_at_ts']) / 3600000) + 1

# 5. Formule de Time Decay : Clics / Âge
popularity_advanced['time_decay_score'] = popularity_advanced[count_col] / popularity_advanced['age_hours']

# 6. On trie par ce nouveau score
popularity_advanced = popularity_advanced.sort_values('time_decay_score', ascending=False)

# --- Affichage et comparaison ---
print(f"\nTop 5 - Popularité Classique (Basée sur {count_col}) :")
display(popularity_df.head(5))

print("\nTop 5 - Popularité Avancée (Time Decay) :")
# On n'affiche que les colonnes intéressantes
display(popularity_advanced[['click_article_id', count_col, 'age_hours', 'time_decay_score']].head(5))

# 7. Sauvegarde pour le benchmark futur
# S'assure que le dossier existe
import os
os.makedirs('../../Generated/Popularity_data', exist_ok=True)

popularity_advanced.to_parquet('../../Generated/Popularity_data/articles_popularity_time_decay.parquet')
print("\n✅ Popularité Time Decay calculée et sauvegardée !")
```

# Matrice user–item et sparsité


## Affichage d'une matrice exemple

📖 Définitions fondamentales

**1. La Matrice User-Item (Utilisateur-Article)**
C'est un tableau à double entrée fondamental dans les systèmes de recommandation. Chaque ligne représente un utilisateur unique, et chaque colonne représente un article unique du catalogue. À l'intersection des deux, on inscrit une valeur (ex: `1` s'il y a eu un clic, `0` sinon). Elle permet de modéliser mathématiquement l'historique complet des intéractions.

**2. La Sparsité (Densitié de la matrice)**
La sparsité désigne le pourcentage de « cases vides » dans cette matrice. Dans la vraie vie, un utilisateur ne lira qu'une infime fraction des articles disponibles. Si une matrice a une sparsité de 99%, cela signifie que 99% des combinaisons possibles *Utilisateur/Article* n'ont jamais eu lieu. Une sparsité trop élevée empêche le bon fonctionnement des algorithmes classiques basés sur le comportement (comme le Filtrage Collaboratif), car il n'y a pas assez de points communs comportementaux entre les utilisateurs.

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import numpy as np

print("📈 Visualisation 4: Sparsité des données")

# Calcul de la vraie sparsité globale de manière mathématique sans exploser la RAM
total_users = clicks_df['user_id'].nunique()
total_items = clicks_df['click_article_id'].nunique()
total_cells = total_users * total_items
non_zero_cells = len(clicks_df[['user_id', 'click_article_id']].drop_duplicates())
sparsity = (1 - (non_zero_cells / total_cells)) * 100

# Création d'un petit échantillon 50x50 pour l'illustration de la heatmap
top_articles = clicks_df['click_article_id'].value_counts().head(50).index
top_users = clicks_df['user_id'].value_counts().head(50).index

sample_df = clicks_df[(clicks_df['user_id'].isin(top_users)) & (clicks_df['click_article_id'].isin(top_articles))]
sample_matrix = sample_df.groupby(['user_id', 'click_article_id']).size().unstack(fill_value=0)
sample_matrix = sample_matrix.reindex(index=top_users, columns=top_articles, fill_value=0)

labels = ['Cellules non-nulles', 'Cellules nulles (sparsité)']
sizes = [non_zero_cells, total_cells - non_zero_cells]
colors = ['#7367F0', '#E2E2E2']

fig = make_subplots(
    rows=1, cols=2,
    specs=[[{"type": "xy"}, {"type": "domain"}]],
    subplot_titles=(
        'Heatmap user-article (échantillon 50x50)', 
        f'Sparsité de la matrice ({sparsity:.2f}% vides)'
    ),
    horizontal_spacing=0.15
)

fig.add_trace(
    go.Heatmap(
        z=sample_matrix.values,
        colorscale=[[0, '#E2E2E2'], [1, '#7367F0']],
        showscale=True,
        colorbar=dict(title='Nombre de clics', x=0.42)
    ),
    row=1, col=1
)

fig.add_trace(
    go.Pie(
        labels=labels, 
        values=sizes, 
        marker_colors=colors,
        textinfo='label+percent',
        hole=0.4
    ),
    row=1, col=2
)

fig.update_xaxes(title_text='Articles (échantillon)', row=1, col=1)
fig.update_yaxes(title_text='Utilisateurs (échantillon)', row=1, col=1)

fig.update_layout(height=500, width=1100, template="plotly_white", showlegend=False)
fig.show()

print(f"📊 Métriques de sparsité globales:")
print(f"- Utilisateurs uniques: {total_users:,}")
print(f"- Articles uniques: {total_items:,}")
print(f"- Cellules totales (matrice globale): {total_cells:,}")
print(f"- Cellules non-nulles (interactions uniques): {non_zero_cells:,}")
print(f"- Sparsité globale: {sparsity:.2f}%")

print(f"\n🎯 Explication de la sparsité dans les systèmes de recommandation :")
print("La sparsité représente le pourcentage d'interactions non existantes (utilisateurs n'ayant pas lu un article spécifique).")
print(f"Une sparsité de {sparsity:.2f}% indique qu'il est difficile d'utiliser uniquement du Collaborative Filtering (filtrage collaboratif)")
print("et qu'une approche Content-Based (basée sur les métadonnées et embeddings des articles) est souvent nécessaire pour pallier au cold start.")
```

## Sauvegarde de la matrice pour le collaborative approche

```python
import scipy.sparse as sparse
import pickle

print("💾 Sauvegarde de la matrice creuse User-Item pour le Collaborative Filtering...")

# Création des index numériques continus (ALS nécessite des index continus)
interactions_df = clicks_df[['user_id', 'click_article_id']].copy()
interactions_df['clicks'] = 1
interactions_df = interactions_df.groupby(['user_id', 'click_article_id']).sum().reset_index()

users = interactions_df['user_id'].astype('category')
items = interactions_df['click_article_id'].astype('category')

user_codes = users.cat.codes
item_codes = items.cat.codes
clicks = interactions_df['clicks'].astype(float)

# Matrice Item-User (préférée pour ALS implicit)
sparse_item_user = sparse.csr_matrix((clicks, (item_codes, user_codes)))
sparse.save_npz('../../Generated/sparse_item_user_matrix.npz', sparse_item_user)

# Sauvegarde des dictionnaires de correspondances
user_mapping = dict(enumerate(users.cat.categories))
item_mapping = dict(enumerate(items.cat.categories))

with open('../../Generated/Mapping_data/user_mapping.pkl', 'wb') as f:
    pickle.dump(user_mapping, f)
with open('../../Generated/Mapping_data/item_mapping.pkl', 'wb') as f:
    pickle.dump(item_mapping, f)

print(f"✅ Matrice sauvegardée dans Generated/sparse_item_user_matrix.npz ({sparse_item_user.shape})")
```
